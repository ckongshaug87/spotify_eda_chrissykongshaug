## Histogram to show variance of values

```python
sns.boxplot(x=df['danceability'])
```

```python
columns_to_corr = [
    'popularity', 'duration_ms', 'danceability', 'energy', 
    'key', 'loudness', 'mode', 'speechiness', 
    'acousticness', 'instrumentalness', 'liveness', 
    'valence', 'tempo', 'time_signature'
]

df_corr = df[columns_to_corr].corr(method='pearson', numeric_only=True)

df_corr
```

```python
fig, ax = plt.subplots(figsize=(15,15))
sns.heatmap(df_corr, annot=True, cmap='mako')
```

# The pair plot helps show the variation in the data

```python
sns.pairplot(df[['acousticness', 'energy']], height=5, aspect=1.2)
```

```python
df_sample=df[['acousticness', 'energy']].sample(n=5000)
sns.pairplot(df[['acousticness', 'energy']], height=5, aspect=1.5, diag_kind='kde')
```

## Digging further into the data. Looking at the data by track_genre. Changing the pair plot for further analysis

```python
df['track_genre'].value_counts()

df_genres = df[df['track_genre'].isin(['pop', 'folk', 'emo', 'punk-rock', 'electronic'])]

df_sample = df_genres.sample(n=5000, random_state=1)

sns.pairplot( df_sample[['acousticness', 'energy', 'track_genre']],  height=5, aspect=1.5, kind='kde', hue='track_genre' )
```
